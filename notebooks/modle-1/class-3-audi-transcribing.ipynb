```python
# import
import sys
import threading
from time import sleep
```

```python
def show_timer(duration):
    for i in range(duration):
        sys.stdout.write(f"\r⏳ Recording... {i+1} seconds")
        sys.stdout.flush()
        sleep(1)
    print()

def record_voice(seconds: int) -> None:
    print(f"Starting voice recording for {seconds} seconds...")
    thread = threading.Thread(target=show_timer, args=(seconds,))
    thread.start()
    # This is a vital piece of code
    # This signals thread to wait until the cell execution is completed
    # before concluding the cell execution
    thread.join()

def run()  -> None:
    try:
        seconds = int(input("For how many seconds do you want to record the voice? "))
        record_voice(seconds)
    except ValueError:
        print("Invalid input. Please enter a valid number of seconds.")

if __name__ == "__main__":
    run()
```
