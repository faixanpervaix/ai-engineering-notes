```python
# import
import sys
import threading
from time import sleep
```

```python
def show_progress(seconds: int) -> None:
    for i in range(seconds):
        sys.stdout.write(f"\rRecording... {i + 1}/{seconds} seconds")
        sys.stdout.flush()
        sleep(1)
    print("\nVoice recording completed!")

def record_voice(seconds: int) -> None:
    print(f"Starting voice recording for {seconds} seconds...")
    threading.Thread(target=show_progress, args=(seconds,)).start()

def run() -> None:
    try:
        seconds = int(input("For how many seconds do you want to record the voice? "))
        record_voice(seconds)
    except ValueError:
        print("Invalid input. Please enter a valid number of seconds.")

if __name__ == "__main__":
    run()
```
