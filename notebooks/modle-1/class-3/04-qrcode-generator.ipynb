```python
# import packages
import qrcode
import qrcode.image.svg
from PIL import Image
```

```python
# qrcode data
qrcode_data: str = "https://faixanpervaix.github.io/ai-engineering-notes"
```

## 1 - Basic

```python
def basic_qr_code() -> None:
    try:
        qrcode.make(qrcode_data).save("basic_qrcode.png")
    except Exception as e:
        print(f"Error: {e}")
    else:
        print("QR code generated and saved as 'basic_qrcode.png'")

basic_qr_code()
```

## 2 - Custom Colors & Sizing

```python
def custom_qr_code() -> None:
    try:
        qr = qrcode.QRCode(
            version=1, # controls the size of the QR code, higher number means bigger code (1 to 40)
            error_correction=qrcode.constants.ERROR_CORRECT_L, # controls the error correction level (L, M, Q, H)
            box_size=10,
            border=4,
        )
        qr.add_data(qrcode_data)
        qr.make(fit=True)
        img = qr.make_image(fill_color="#073B4C", back_color="#FFD166")
        img.save("custom_color_and_sizing_qrcode.png")
    except Exception as e:
        print(f"Error: {e}")
    else:
        print("Custom QR code generated and saved as 'custom_color_and_sizing_qrcode.png'")

custom_qr_code()
```

## 3 - Adding Logo in QR

```python
def qrcode_with_logo() -> None:
    try:
        qr = qrcode.QRCode(
            version=5,
            error_correction=qrcode.constants.ERROR_CORRECT_H,
            box_size=10,
            border=4,
        )
        qr.add_data(qrcode_data)
        qr.make(fit=True)
        img = qr.make_image(fill_color="#073B4C", back_color="#FFD166").convert('RGB')

        logo = Image.open("gh_logo.jpg")
        logo = logo.resize((img.size[0] // 4, img.size[1] // 4))
        pos = ((img.size[0] - logo.size[0]) // 2, (img.size[1] - logo.size[1]) // 2)
        img.paste(logo, pos)
        img.save("logo_qrcode.png")
    except Exception as e:
        print(f"Error: {e}")
    else:
        print("Custom QR code generated and saved as 'logo_qrcode.png'")

qrcode_with_logo()
```

## 4 - SVG QR Code

```python
def create_svg_qr_code() -> None:
    try:
        factory = qrcode.image.svg.SvgImage
        img = qrcode.make(qrcode_data, image_factory=factory)
        img.save("vector_qr.svg")
    except Exception as e:
        print(f"Error: {e}")
    else:
        print("SVG QR code generated and saved as 'vector_qr.svg'")

create_svg_qr_code()
```

## 5 - Transparent Background QR Code

```python
def create_qr_code_with_transparent_bg() -> None:
    try:
        qr = qrcode.QRCode(
            version=1,
            error_correction=qrcode.constants.ERROR_CORRECT_L,
            box_size=10,
            border=4,
        )
        qr.add_data(qrcode_data)
        qr.make(fit=True)
        img = qr.make_image(fill_color="#073B4C", back_color="transparent").convert('RGBA')
        img.save("transparent_qr.png")
    except Exception as e:
        print(f"Error: {e}")
    else:
        print("Custom QR code generated and saved as 'transparent_qr.png'")

create_qr_code_with_transparent_bg()
```

## 6 - Wifi QR Code

```python
def create_wifi_qrcode(ssid, password, auth_type="WPA", hidden=False):
    hidden_str = "true" if hidden else "false"
    wifi_data = f"WIFI:T:{auth_type};S:{ssid};P:{password};H:{hidden_str};;"
    try:
        img = qrcode.make(wifi_data)
        img.save("wifi_qr.png")
    except Exception as e:
        print(f"Error: {e}")
    else:
        print(f"WiFi QR code for '{ssid}' generated successfully and saved as 'wifi_qr.png'.")

create_wifi_qrcode(
    ssid="YourNetworkName", 
    password="YourSecurePassword", 
    auth_type="WPA",  # Use "WPA", "WEP", or "nopass" (for no password)
    hidden=False
)
```

## 7 - VCard QR Code

```python
def create_vcard_qr(
    first_name : str,
    last_name : str,
    phone : str,
    email : str,
    company : str,
    title : str
) -> None:
    # Construct the vCard string (Version 3.0 is standard for mobile devices)
    vcard_data = (
        "BEGIN:VCARD\n"
        "VERSION:3.0\n"
        f"N:{last_name};{first_name}\n"
        f"FN:{first_name} {last_name}\n"
        f"ORG:{company}\n"
        f"TITLE:{title}\n"
        f"TEL;TYPE=WORK,VOICE:{phone}\n"
        f"EMAIL:{email}\n"
        "END:VCARD"
    )
    
    try:
        img = qrcode.make(vcard_data)
        img.save("vcard_qr.png")
        print("vCard QR code generated successfully.")
    except Exception as e:
        print(f"Error: {e}")
    else:
        print("vCard QR code generated and saved as 'vcard_qr.png'.")

create_vcard_qr(
    first_name="Jane",
    last_name="Doe",
    phone="+1234567890",
    email="jane.doe@example.com",
    company="Tech Corp",
    title="Backend Developer"
)
```
