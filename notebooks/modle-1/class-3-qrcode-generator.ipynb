```python
# import packages
import qrcode
import qrcode.image.svg
from PIL import Image
```

```python
# qrcode data
qrcode_data: str = "https://faixanpervaix.github.io/ai-engineering-notes"
```

## 1 - Basic

```python
try:
    qrcode.make(qrcode_data).save("basic_qrcode.png")
except Exception as e:
    print(f"Error: {e}")
else:
    print("QR code generated and saved as 'basic_qrcode.png'")
```

## 2 - Custom Colors & Sizing

```python
try:
    qr = qrcode.QRCode(
        version=1, # controls the size of the QR code, higher number means bigger code (1 to 40)
        error_correction=qrcode.constants.ERROR_CORRECT_L, # controls the error correction level (L, M, Q, H)
        box_size=10,
        border=4,
    )
    qr.add_data(qrcode_data)
    qr.make(fit=True)
    img = qr.make_image(fill_color="#073B4C", back_color="#FFD166")
    img.save("custom_color_and_sizing_qrcode.png")
except Exception as e:
    print(f"Error: {e}")
else:
    print("Custom QR code generated and saved as 'custom_color_and_sizing_qrcode.png'")
```

## 3 - Adding Logo in QR

```python
try:
    qr = qrcode.QRCode(
        version=5,
        error_correction=qrcode.constants.ERROR_CORRECT_H,
        box_size=10,
        border=4,
    )
    qr.add_data(qrcode_data)
    qr.make(fit=True)
    img = qr.make_image(fill_color="#073B4C", back_color="#FFD166").convert('RGB')

    logo = Image.open("gh_logo.jpg")
    logo = logo.resize((img.size[0] // 4, img.size[1] // 4))
    pos = ((img.size[0] - logo.size[0]) // 2, (img.size[1] - logo.size[1]) // 2)
    img.paste(logo, pos)
    img.save("logo_qrcode.png")
except Exception as e:
    print(f"Error: {e}")
else:
    print("Custom QR code generated and saved as 'logo_qrcode.png'")
```

## 4 - SVG QR Code

```python
try:
    factory = qrcode.image.svg.SvgImage
    img = qrcode.make(qrcode_data, image_factory=factory)
    img.save("vector_qr.svg")
except Exception as e:
    print(f"Error: {e}")
else:
    print("SVG QR code generated and saved as 'vector_qr.svg'")
```

## 5 - Transparent Background QR Code

```python
try:
    qr = qrcode.QRCode(
        version=1,
        error_correction=qrcode.constants.ERROR_CORRECT_L,
        box_size=10,
        border=4,
    )
    qr.add_data(qrcode_data)
    qr.make(fit=True)
    img = qr.make_image(fill_color="#073B4C", back_color="transparent").convert('RGBA')
    img.save("transparent_qr.png")
except Exception as e:
    print(f"Error: {e}")
else:
    print("Custom QR code generated and saved as 'transparent_qr.png'")
```
