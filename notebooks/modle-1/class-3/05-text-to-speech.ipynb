```python
# imports
import os
from pathlib import Path
from typing import List, Optional

import pyttsx3
from gtts import gTTS
from pyttsx3.engine import Engine
from pyttsx3.voice import Voice
```

```python
# constants
CURRENT_DIR: Path = Path.cwd()
SPEECH_TEXT_FILE: Path = CURRENT_DIR / "speech.txt"
```

```python
def get_file_content() -> Optional[str]:
    try:
        with open(SPEECH_TEXT_FILE, "r") as file:
            return file.read().strip()
    except FileNotFoundError:
        print("\nEnter the text you want to convert to speech:")
        text_to_speech = input(">> ").strip()
        return text_to_speech if text_to_speech else "You have provided an empty input."
    except Exception:
        return "Error occurred while reading the file."
```

```python
def text_to_speech_using_pyttsx3() -> None:
    try:
        engine: Engine = pyttsx3.init()
        voices: List[Voice] = engine.getProperty("voices")

        for i, voice in enumerate(voices):
            serial_number = i + 1
            print(f"{serial_number}: {voice.name}")

            if serial_number >= 5:
                break

        selected_voice: str = input(
            "Select a voice by entering the corresponding number: "
        )

        if selected_voice.isdigit() and 1 <= int(selected_voice) <= 5:
            engine.setProperty("voice", voices[int(selected_voice) - 1].id)
            engine.setProperty("rate", 160)  # Set speech rate (optional)
            engine.setProperty("volume", 1.0)  # Set volume (optional)

        engine.say(get_file_content())
        engine.runAndWait()
    except Exception as e:
        print(f"Error: {e}")
        raise e
```

#### Using gTTS

```python
def text_to_speech_using_gTTS(text: str) -> None:
    try:
        print(
            "Choose language (Google TTS supports many codes, e.g., 'en' for English, 'ur' for Urdu, 'es' for Spanish):"
        )
        lang = input("Enter language code (default 'en'): ") or "en"

        tts = gTTS(text=get_file_content(), lang=lang)

        output_file = "output.mp3"
        tts.save(output_file)
        print(f"Saved speech as '{output_file}'")

        if os.name == "nt":
            os.system(f"start {output_file}")

        elif os.name == "posix":
            os.system(f"afplay {output_file}")
    except Exception as e:
        print("Error:", e)
```

```python
def main():
    print("Please choose the text-to-speech engine:")
    print("1. pyttsx3")
    print("2. gTTS")
    choice = input("Enter your choice (1 or 2): ")
    if choice == "1":
        text_to_speech_using_pyttsx3()
    elif choice == "2":
        text_to_speech_using_gTTS(get_file_content())
    else:
        print("Invalid choice.")
```

```python
if __name__ == "__main__":
    main()
```
