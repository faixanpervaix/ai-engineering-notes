```python
# import packages
import pandas as pd
import requests as req
from requests.exceptions import HTTPError
from json import JSONDecodeError
from IPython.display import display # this is for displaying dataframes in a nice format in Jupyter notebooks in vs code if nicely formatted output is not showing up.
```

## 1 - Get Products Data Using [FakeStore Api](https://fakestoreapi.com/products)

**Endpoint:** https://fakestoreapi.com/products

```python
fake_store_api: str = "https://fakestoreapi.com/products"
df: pd.DataFrame | None = None
try:
    response = req.get(fake_store_api)
    data = response.json()
    df = pd.DataFrame(data)
    display(df.head())
except HTTPError as http_err:
    print(f"HTTP error occurred: {http_err}")
except JSONDecodeError as json_err:
    print(f"JSON decode error occurred: {json_err}")
except Exception as err:
    print(f"An error occurred: {err}")
```

## 2 - Generate CSV

```python
if df is not None:
    df.to_csv("products.csv", index=False)
```
